```python
from pathlib import Path
import sys

import numpy as np
import pandas as pd

import torch
from torch.utils.data import DataLoader
import torch.nn as nn
import torch.nn.functional as F

import matplotlib.pyplot as plt

from transformers import CLIPVisionModel, CLIPImageProcessor

# --- find project root ---

def find_project_root(start: Path) -> Path:
    cur = start.resolve()
    for p in [cur] + list(cur.parents):
        if (p / "src").is_dir() and (p / "data").is_dir():
            return p
    raise RuntimeError(f"Could not find project root starting from {start}")

cwd = Path.cwd()
project_root = find_project_root(cwd)
print("CWD         :", cwd)
print("PROJECT_ROOT:", project_root)

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
print("project_root in sys.path:", str(project_root) in sys.path)

from src.clip_domain_datasets import ClipDomainDataset, clip_collate_fn

META_DIR = project_root / "data" / "processed" / "metadata"
CSV_PATH = META_DIR / "domain_classification_sentences.csv"

CKPT_CLIP_A3 = project_root / "checkpoints" / "clip_domain_A3_clip.pt"
CLIP_MODEL_NAME = "openai/clip-vit-base-patch32"

print("\nCSV_PATH     :", CSV_PATH)
print("CKPT_CLIP_A3 :", CKPT_CLIP_A3)

df = pd.read_csv(CSV_PATH)
print("\nCSV shape:", df.shape)
print("hf_split values:", df["hf_split"].unique())
print("domain_label counts:\n", df["domain_label"].value_counts())

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("\nUsing device:", device)

class ClipDomainClassifier(nn.Module):
    def __init__(self, clip_vision: CLIPVisionModel, hidden_size: int, num_classes: int = 2):
        super().__init__()
        self.clip_vision = clip_vision
        self.classifier = nn.Linear(hidden_size, num_classes)

    def forward(self, pixel_values: torch.Tensor) -> torch.Tensor:
        outputs = self.clip_vision(pixel_values=pixel_values)
        feats = outputs.pooler_output
        logits = self.classifier(feats)
        return logits
```

```python
print(f"Loading CLIP vision model: {CLIP_MODEL_NAME}")
clip_processor = CLIPImageProcessor.from_pretrained(CLIP_MODEL_NAME)
clip_vision = CLIPVisionModel.from_pretrained(CLIP_MODEL_NAME).to(device)

for p in clip_vision.parameters():
    p.requires_grad = False

hidden_size = clip_vision.config.hidden_size
print("CLIP hidden_size:", hidden_size)

# A3: normalize_all=True for ALL splits, including test
test_ds_clip_A3 = ClipDomainDataset(
    csv_path=CSV_PATH,
    project_root=project_root,
    split="test",
    clip_processor=clip_processor,
    augment=False,
    photo_augment=False,
    normalize_all=True,
)

test_loader_clip_A3 = DataLoader(
    test_ds_clip_A3,
    batch_size=64,
    shuffle=False,
    collate_fn=clip_collate_fn,
)

model_clip_A3 = ClipDomainClassifier(
    clip_vision=clip_vision,
    hidden_size=hidden_size,
    num_classes=2,
).to(device)

print("\nModel:", model_clip_A3.__class__.__name__)

if CKPT_CLIP_A3.exists():
    state = torch.load(CKPT_CLIP_A3, map_location=device)
    model_clip_A3.load_state_dict(state)
    print("Loaded A3 CLIP checkpoint from:", CKPT_CLIP_A3)
else:
    print("WARNING: CLIP A3 checkpoint not found, using random weights!")

model_clip_A3.eval()
```

```python
all_true = []
all_pred = []
all_prob_fake = []
all_sources = []
all_idxs = []
all_texts = []

criterion = nn.CrossEntropyLoss()

model_clip_A3.eval()

with torch.no_grad():
    total_loss = 0.0
    total_samples = 0

    for batch in test_loader_clip_A3:
        pixel_values = batch["pixel_values"].to(device)
        labels = batch["labels"].to(device)

        logits = model_clip_A3(pixel_values)
        loss = criterion(logits, labels)

        probs = F.softmax(logits, dim=1)
        pred = logits.argmax(dim=1)

        total_loss += loss.item() * pixel_values.size(0)
        total_samples += pixel_values.size(0)

        all_true.extend(labels.cpu().numpy().tolist())
        all_pred.extend(pred.cpu().numpy().tolist())
        all_prob_fake.extend(probs[:, 1].cpu().numpy().tolist())

        all_sources.extend(batch["sources"])
        all_idxs.extend(batch["idxs"])
        all_texts.extend(batch["texts"])

test_loss_A3 = total_loss / total_samples
test_acc_A3 = (np.array(all_true) == np.array(all_pred)).mean()

test_results_clip_A3 = pd.DataFrame({
    "true_label": all_true,
    "pred_label": all_pred,
    "prob_fake": all_prob_fake,
    "source": all_sources,
    "idx": all_idxs,
    "text": all_texts,
})

print("test_results_clip_A3 shape:", test_results_clip_A3.shape)
display(test_results_clip_A3.head())

print(f"\nCLIP A3 test loss: {test_loss_A3:.4f}")
print(f"CLIP A3 test accuracy: {test_acc_A3*100:.2f}%")
```

```python
num_classes = 2
cm_clip_A3 = np.zeros((num_classes, num_classes), dtype=int)

for t, p in zip(test_results_clip_A3["true_label"], test_results_clip_A3["pred_label"]):
    cm_clip_A3[t, p] += 1

print("Confusion matrix (CLIP A3; rows = true, cols = pred):")
print("       pred=0    pred=1")
print(f"true=0   {cm_clip_A3[0,0]:6d}   {cm_clip_A3[0,1]:6d}")
print(f"true=1   {cm_clip_A3[1,0]:6d}   {cm_clip_A3[1,1]:6d}")

acc_iam_A3   = cm_clip_A3[0,0] / cm_clip_A3[0].sum() if cm_clip_A3[0].sum() > 0 else 0.0
acc_emuru_A3 = cm_clip_A3[1,1] / cm_clip_A3[1].sum() if cm_clip_A3[1].sum() > 0 else 0.0

print("\nPer-class accuracy (CLIP A3):")
print(f"  Class 0 (IAM / genuine): {acc_iam_A3*100:.2f}%")
print(f"  Class 1 (Emuru / fake) : {acc_emuru_A3*100:.2f}%")
```

```python
from IPython.display import display
from PIL import Image

mis_clip_A3 = test_results_clip_A3[
    test_results_clip_A3["true_label"] != test_results_clip_A3["pred_label"]
].copy()
print("Number of misclassified test samples (CLIP A3):", len(mis_clip_A3))

df_test = df[df["hf_split"] == "test"].copy()

mis_merged_clip_A3 = mis_clip_A3.merge(
    df_test[["idx", "source", "filepath", "domain_label"]],
    on=["idx", "source"],
    how="left",
    suffixes=("", "_orig"),
)

print("Misclassified merged shape (A3):", mis_merged_clip_A3.shape)

if len(mis_merged_clip_A3) == 0:
    print("\nNo misclassifications for CLIP A3 🎉")
    print("Showing borderline examples instead.")
    test_results_clip_A3["dist_to_0_5"] = (test_results_clip_A3["prob_fake"] - 0.5).abs()
    rows_to_show_A3 = test_results_clip_A3.sort_values("dist_to_0_5").head(8).merge(
        df_test[["idx", "source", "filepath", "domain_label"]],
        on=["idx", "source"],
        how="left",
        suffixes=("", "_orig"),
    )
else:
    rows_to_show_A3 = mis_merged_clip_A3.head(8)

print("\nRows to show:", len(rows_to_show_A3))

for i, row in rows_to_show_A3.iterrows():
    print("=" * 80)
    print(f"idx: {row['idx']} | source: {row['source']}")
    print(f"text: {repr(row['text'])}")
    print(f"true_label    : {row['true_label']} (0=IAM/genuine, 1=Emuru/fake)")
    print(f"pred_label    : {row['pred_label']}")
    print(f"prob_fake (p(class=1)) = {row['prob_fake']:.4f}")
    print(f"filepath: {row['filepath']}")

    img_path = project_root / row["filepath"]

    try:
        img = Image.open(img_path).convert("L")
        display(img)
    except Exception as e:
        print("  Could not load/display image:", e)
```

```python
from sklearn.manifold import TSNE

MAX_POINTS = 1500

all_feats = []
all_true = []
all_pred = []
all_prob_fake = []
all_sources = []
all_idxs = []
all_texts = []

model_clip_A3.eval()
with torch.no_grad():
    for batch in test_loader_clip_A3:
        pixel_values = batch["pixel_values"].to(device)
        labels = batch["labels"]
        sources = batch["sources"]
        idxs = batch["idxs"]
        texts = batch["texts"]

        logits = model_clip_A3(pixel_values)
        probs = F.softmax(logits, dim=1)
        pred = logits.argmax(dim=1)

        outputs = model_clip_A3.clip_vision(pixel_values=pixel_values)
        feats = outputs.pooler_output

        all_feats.append(feats.cpu().numpy())
        all_true.extend(labels.cpu().numpy().tolist())
        all_pred.extend(pred.cpu().numpy().tolist())
        all_prob_fake.extend(probs[:, 1].cpu().numpy().tolist())
        all_sources.extend(sources)
        all_idxs.extend(idxs)
        all_texts.extend(texts)

        if len(all_true) >= MAX_POINTS:
            break

X_A3 = np.concatenate(all_feats, axis=0)
N_A3 = min(len(all_true), MAX_POINTS)
X_A3 = X_A3[:N_A3]
y_true_A3 = np.array(all_true[:N_A3])
y_pred_A3 = np.array(all_pred[:N_A3])
prob_fake_A3 = np.array(all_prob_fake[:N_A3])
sources_sub_A3 = all_sources[:N_A3]
idxs_sub_A3 = all_idxs[:N_A3]
texts_sub_A3 = all_texts[:N_A3]

print("Using N =", N_A3, "points for t-SNE (A3)")
print("Feature matrix shape:", X_A3.shape)

tsne_A3 = TSNE(
    n_components=2,
    init="random",
    learning_rate="auto",
    random_state=42,
    perplexity=30,
)
X_emb_A3 = tsne_A3.fit_transform(X_A3)
print("t-SNE embedding shape (A3):", X_emb_A3.shape)

vis_df_clip_A3 = pd.DataFrame({
    "true_label": y_true_A3,
    "pred_label": y_pred_A3,
    "prob_fake": prob_fake_A3,
    "source": sources_sub_A3,
    "idx": idxs_sub_A3,
    "text": texts_sub_A3,
})

mis_mask_A3 = (vis_df_clip_A3["true_label"] != vis_df_clip_A3["pred_label"]).to_numpy()

vis_df_clip_A3["dist_to_0_5"] = (vis_df_clip_A3["prob_fake"] - 0.5).abs()
borderline_idx_A3 = vis_df_clip_A3["dist_to_0_5"].nsmallest(50).index
borderline_mask_A3 = np.zeros(len(vis_df_clip_A3), dtype=bool)
borderline_mask_A3[borderline_idx_A3.to_numpy()] = True

iam_mask_A3 = (vis_df_clip_A3["true_label"].to_numpy() == 0)
emu_mask_A3 = (vis_df_clip_A3["true_label"].to_numpy() == 1)

plt.figure(figsize=(8, 6))

plt.scatter(
    X_emb_A3[iam_mask_A3, 0],
    X_emb_A3[iam_mask_A3, 1],
    s=4,
    alpha=0.4,
    label="IAM (genuine)",
)
plt.scatter(
    X_emb_A3[emu_mask_A3, 0],
    X_emb_A3[emu_mask_A3, 1],
    s=4,
    alpha=0.4,
    label="Emuru (fake)",
)

if mis_mask_A3.any():
    plt.scatter(
        X_emb_A3[mis_mask_A3, 0],
        X_emb_A3[mis_mask_A3, 1],
        s=40,
        edgecolors="k",
        facecolors="none",
        linewidths=1.0,
        label="Misclassified",
    )

plt.scatter(
    X_emb_A3[borderline_mask_A3, 0],
    X_emb_A3[borderline_mask_A3, 1],
    s=40,
    marker="x",
    linewidths=1.0,
    label="Borderline (prob≈0.5)",
)

plt.title("t-SNE of CLIP A3 test features (subset)")
plt.xlabel("t-SNE dim 1")
plt.ylabel("t-SNE dim 2")
plt.legend()
plt.tight_layout()
plt.show()
```
